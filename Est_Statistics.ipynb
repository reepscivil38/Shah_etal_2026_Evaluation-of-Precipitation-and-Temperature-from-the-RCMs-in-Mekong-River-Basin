```python
##################################################################################################
### Calculate spatically explicit corr, CRMS (Centered Root Mean Square) & RMSE: Preci & Temp  ###
##################################################################################################

import pandas as pd
import numpy as np

# define the input files
input_f = [[[df_day_cord_preci,  # preci [0,0]
           df_mon_cord_preci,          #[0,1]
           df_yr_cord_preci,
           df_cord_preci],
           [df_day_cord_JJASO_preci,  # preci [0,0]
           df_mon_cord_JJASO_preci,          #[0,1]
           df_yr_cord_JJASO_preci,
           df_cord_JJASO_preci]],
           [[df_day_cord_temp,  # preci [0,0]
           df_mon_cord_temp,          #[0,1]
           df_yr_cord_temp,
           df_cord_temp],
           [df_day_cord_JJASO_temp,  # preci [0,0]
           df_mon_cord_JJASO_temp,          #[0,1]
           df_yr_cord_JJASO_temp,
           df_cord_JJASO_temp]]]
var_type = ['preci', 'temp']
season_type = ['All', 'JJASO']
TScale = ['Day', 'Mon', 'Yr', '18Yr']
#target_col = 'APHRO'
#feature_cols = ['APHRO', 'ERA-I', 'HIRHAM (EAS)', 'HadRM3P (EAS)','RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)', 'RCA4 (WAS)', 'CCLM (EAS)']

# define the function to calculate correlation by group
def correlation_by_group(df, group_col, target_col, feature_cols):
    """
    Calculates the correlation between a target column and multiple feature columns,
    grouped by another column.

    Args:
        df (pd.DataFrame): The input DataFrame.
        group_col (str): The name of the column to group by.
        target_col (str): The name of the target column.
        feature_cols (list): A list of column names to calculate correlation with the target column.

    Returns:
        pd.DataFrame: A DataFrame containing the correlation coefficients for each group and feature column.
    """
    correlations = df.groupby(group_col).apply(lambda x: x[feature_cols].corrwith(x[target_col]))
    return correlations

# calculate CRMS, correlation for each grid
for k in range(2):
    for i in range(2):  # 0, 1
        for j in range(3): # 0, 1, 2, 3  # only execute for 0,1,2, because 3 will yield sigular stdev (one data point for each grid)
            df_in = input_f[k][i][j].copy() #instead of referencing the original dataframe, we create a copy to avoid modifying it

            # Pearson correlation coefficient calculation
            group_col = 'Coord'
            target_col = 'APHRO'
            feature_cols = ['APHRO','ERA-I', 'HIRHAM (EAS)', 'HadRM3P (EAS)','RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)', 'RCA4 (WAS)', 'CCLM (EAS)']

            correlations_df = correlation_by_group(df_in, group_col, target_col, feature_cols)
            
            # stdandard deviation calculation
            std_by_group = df_in.groupby(group_col)[feature_cols].std()
            
            # centered root mean square (CRMS) calculation
            ref_std_by_group = std_by_group[target_col]
            cross_by_group = correlations_df.transform(lambda x: x*ref_std_by_group, axis=0)
            cross_by_group = cross_by_group*std_by_group
            crms_by_group = std_by_group[feature_cols]**2
            ref_std_by_group = std_by_group[target_col]**2
            crms_by_group = crms_by_group.transform(lambda x: x + ref_std_by_group, #- 2*cross_by_group, 
                                                    axis=0) 
            crms_by_group = crms_by_group.transform(lambda x: np.sqrt(x - 2*cross_by_group), 
                                                    axis=0)                                    
                     
            correlations_df['Var'] = var_type[k]
            correlations_df['Season'] = season_type[i]
            correlations_df['TScale'] = TScale[j]
            
            crms_by_group['Var'] = var_type[k]
            crms_by_group['Season'] = season_type[i]
            crms_by_group['TScale'] = TScale[j]

            # Calculate the root mean squared error (RMSE) 
            rmse_by_group = df_in
            feature_cols = ['ERA-I', 'HIRHAM (EAS)', 'HadRM3P (EAS)','RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)', 'RCA4 (WAS)', 'CCLM (EAS)']
            
            for col in feature_cols:
                rmse_by_group[col] = (rmse_by_group[col] - rmse_by_group[target_col])**2
            rmse_by_group = np.sqrt(rmse_by_group.groupby(group_col)[feature_cols].mean())   
                  
            rmse_by_group['Var'] = var_type[k]
            rmse_by_group['Season'] = season_type[i]
            rmse_by_group['TScale'] = TScale[j]
            
                        
            if k==0 and i == 0 and j == 0:
                
                df2 = correlations_df   
                df3 = crms_by_group
                df4 = rmse_by_group
                
            else:
                
                df2 = pd.concat([df2,correlations_df])
                df3 = pd.concat([df3,crms_by_group])
                df4 = pd.concat([df4,rmse_by_group])
                
                
df2.to_csv('../Data/corr_281grids.csv', index=True, encoding='utf-8')    
df3.to_csv('../Data/crms_281grids.csv', index=True, encoding='utf-8')  
df4.to_csv('../Data/rmse_281grids.csv', index=True, encoding='utf-8') 

```

```python
##################################################################
### Calculate mean, median, 5% quantile, 95% quantile of 
### bias (Absolute difference) of Preci & Temp                 ###
##################################################################

import pandas as pd
import numpy as np

# define the input files
input_f = [[[df_day_cord_preci,  # preci [0,0]
           df_mon_cord_preci,          #[0,1]
           df_yr_cord_preci,
           df_cord_preci],
           [df_day_cord_JJASO_preci,  # preci [0,0]
           df_mon_cord_JJASO_preci,          #[0,1]
           df_yr_cord_JJASO_preci,
           df_cord_JJASO_preci]],
           [[df_day_cord_temp,  # preci [0,0]
           df_mon_cord_temp,          #[0,1]
           df_yr_cord_temp,
           df_cord_temp],
           [df_day_cord_JJASO_temp,  # preci [0,0]
           df_mon_cord_JJASO_temp,          #[0,1]
           df_yr_cord_JJASO_temp,
           df_cord_JJASO_temp]]]
var_type = ['preci', 'temp']
season_type = ['All', 'JJASO']
TScale = ['Day', 'Mon', 'Yr', '18Yr']

# calculate bias for each grid
for k in range(2):
    for i in range(2):  # 0, 1
        for j in range(3): # 0, 1, 2, 3  # only execute for 0,1,2, because 3 will yield sigular stdev (one data point for each grid)
            bias = input_f[k][i][j].copy() # instead of referencing, create a copy to avoid modifying the original DataFrame
            
            group_col = 'Coord'
            target_col = 'APHRO'
            feature_cols = ['ERA-I', 'HIRHAM (EAS)', 'HadRM3P (EAS)','RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)', 'RCA4 (WAS)', 'CCLM (EAS)']
                       
            # Calculate bias & statistics of bias
            for col in feature_cols:
                bias[col] = bias[col] - bias[target_col]
            MeanB_by_group = bias.groupby(group_col)[feature_cols].mean()
            MedianB_by_group = bias.groupby(group_col)[feature_cols].median()
            q95B_by_group = bias.groupby(group_col)[feature_cols].quantile(0.95) 
            q05B_by_group = bias.groupby(group_col)[feature_cols].quantile(0.05)  
            
                       
            MeanB_by_group['Var'] = var_type[k]
            MeanB_by_group['Season'] = season_type[i]
            MeanB_by_group['TScale'] = TScale[j]
            
            MedianB_by_group['Var'] = var_type[k]
            MedianB_by_group['Season'] = season_type[i]
            MedianB_by_group['TScale'] = TScale[j]
            
            q95B_by_group['Var'] = var_type[k]
            q95B_by_group['Season'] = season_type[i]        
            q95B_by_group['TScale'] = TScale[j]
            
            q05B_by_group['Var'] = var_type[k]  
            q05B_by_group['Season'] = season_type[i]
            q05B_by_group['TScale'] = TScale[j]
                       
            if k==0 and i == 0 and j == 0:
                df1 = MeanB_by_group
                df2 = MedianB_by_group
                df3 = q95B_by_group
                df4 = q05B_by_group
                
            else:
                df1 = pd.concat([df1,MeanB_by_group])
                df2 = pd.concat([df2,MedianB_by_group])
                df3 = pd.concat([df3,q95B_by_group])
                df4 = pd.concat([df4,q05B_by_group])
                              

df1.to_csv('../Data/MeanBias_281grids.csv', index=True, encoding='utf-8') 
df2.to_csv('../Data/MedianBias_281grids.csv', index=True, encoding='utf-8')
df3.to_csv('../Data/q95Bias_281grids.csv', index=True, encoding='utf-8')
df4.to_csv('../Data/q05Bias_281grids.csv', index=True, encoding='utf-8')   
```
