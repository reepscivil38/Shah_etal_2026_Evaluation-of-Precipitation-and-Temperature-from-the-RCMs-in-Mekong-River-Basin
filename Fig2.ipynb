```python
###########################################################################################
### Codes used to generate Fig 2 (A) (B)
###########################################################################################


###########################################################################################
### read data: precipitation & temperature (All-season and JJASO) daily data            ###
###########################################################################################

import pandas as pd


model_var = ['APHRO', 'ERA-I', 
             'HIRHAM (EAS)', 'HadRM3P (EAS)', 'RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)', 'RCA4 (WAS)', 'CCLM (EAS)']

### All-season precipitation ##########################################################################
### daily precipitation
df = pd.read_csv('../Data/precipitation_daily_1990-2007.csv', index_col=0)

### daily precipitation at each grid
df_day_cord_preci = df  

### create monthly precipitation (mm/day) at each grid
group_var = ['Year','Month','Coord']
df_mon_cord_preci = df.groupby(group_var)[model_var].mean()
df_mon_cord_preci = df_mon_cord_preci.reset_index()

### calculate annual precipitation (mm/day) at each grid
group_var = ['Year','Coord']
df_yr_cord_preci = df.groupby(group_var)[model_var].mean()
df_yr_cord_preci = df_yr_cord_preci.reset_index()

### calculate avg precipitation (mm/day) at each grid
group_var = ['Coord']
df_cord_preci = df.groupby(group_var)[model_var].mean()
df_cord_preci = df_cord_preci.reset_index()

### calculate annual precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year']
df_yr_preci = df.groupby(group_var)[model_var].mean()
df_yr_preci = df_yr_preci.reset_index()

### calculate monthly precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year', 'Month']
df_mon_preci = df.groupby(group_var)[model_var].mean()
df_mon_preci = df_mon_preci.reset_index()

### calculate daily precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year', 'Month', 'Day']
df_day_preci = df.groupby(group_var)[model_var].mean()
df_day_preci = df_day_preci.reset_index()
### End of All-season precipitation ##########################################################################

### JJASO, monsoon precipitation #############################################################################
### daily precipitation
df = df.loc[(df['Month'] > 5) & (df['Month'] < 11)]#pd.read_csv('../Data/precipitation_daily_Jun-Oct_1990-2007.csv', index_col=0)

### daily precipitation at each grid
df_day_cord_JJASO_preci = df  

### create monthly precipitation (mm/day) at each grid
group_var = ['Year','Month','Coord']
df_mon_cord_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_mon_cord_JJASO_preci = df_mon_cord_JJASO_preci.reset_index()

### calculate annual precipitation (mm/day) at each grid
group_var = ['Year','Coord']
df_yr_cord_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_yr_cord_JJASO_preci = df_yr_cord_JJASO_preci.reset_index()

### calculate avg precipitation (mm/day) at each grid
group_var = ['Coord']
df_cord_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_cord_JJASO_preci = df_cord_JJASO_preci.reset_index()

### calculate annual precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year']
df_yr_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_yr_JJASO_preci = df_yr_JJASO_preci.reset_index()

### calculate monthly precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year', 'Month']
df_mon_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_mon_JJASO_preci = df_mon_JJASO_preci.reset_index()

### calculate daily precipitation (mm/day) averaged across all 281 grids 
group_var = ['Year', 'Month', 'Day']
df_day_JJASO_preci = df.groupby(group_var)[model_var].mean()
df_day_JJASO_preci = df_day_JJASO_preci.reset_index()
### End of JJASO, monsoon precipitation #############################################################################

### All-season temperature  #############################################################################
### daily temperature
df = pd.read_csv('../Data/temperature_avg_daily_1990-2007.csv', index_col=0)

### daily temperature at each grid
df_day_cord_temp = df  

### create monthly mean temperature at each grid
group_var = ['Year','Month','Coord']
df_mon_cord_temp = df.groupby(group_var)[model_var].mean()
df_mon_cord__temp = df_mon_cord_temp.reset_index()

### calculate annual mean temperature at each grid
group_var = ['Year','Coord']
df_yr_cord_temp = df.groupby(group_var)[model_var].mean()
df_yr_cord_temp = df_yr_cord_temp.reset_index()

### calculate mean temperature at each grid
group_var = ['Coord']
df_cord_temp = df.groupby(group_var)[model_var].mean()
df_cord_temp = df_cord_temp.reset_index()

### calculate annual mean temperature averaged across all 281 grids 
group_var = ['Year']
df_yr_temp = df.groupby(group_var)[model_var].mean()
df_yr_temp = df_yr_temp.reset_index()

### calculate monthly mean temperature averaged across all 281 grids 
group_var = ['Year', 'Month']
df_mon_temp = df.groupby(group_var)[model_var].mean()
df_mon_temp = df_mon_temp.reset_index()

### calculate daily mean temperature averaged across all 281 grids 
group_var = ['Year', 'Month', 'Day']
df_day_temp = df.groupby(group_var)[model_var].mean()
df_day_temp = df_day_temp.reset_index()


### JJASO monsoon mean temperature ##########################################################################
### daily mean temperature
df = df.loc[(df['Month'] > 5) & (df['Month'] < 11)]#pd.read_csv('../Data/temperature_avg_daily_Jun-Oct_1990-2007.csv', index_col=0)

### daily mean temperature at each grid
df_day_cord_JJASO_temp = df  

### create monthly mean temperature at each grid
group_var = ['Year','Month','Coord']
df_mon_cord_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_mon_cord_JJASO_temp = df_mon_cord_JJASO_temp.reset_index()

### calculate annual  mean temperature at each grid
group_var = ['Year','Coord']
df_yr_cord_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_yr_cord_JJASO_temp = df_yr_cord_JJASO_temp.reset_index()

### calculate  mean temperature at each grid
group_var = ['Coord']
df_cord_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_cord_JJASO_temp = df_cord_JJASO_temp.reset_index()

### calculate annual  mean temperature averaged across all 281 grids 
group_var = ['Year']
df_yr_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_yr_JJASO_temp = df_yr_JJASO_temp.reset_index()

### calculate monthly  mean temperature averaged across all 281 grids 
group_var = ['Year', 'Month']
df_mon_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_mon_JJASO_temp = df_mon_JJASO_temp.reset_index()

### calculate daily  mean temperature averaged across all 281 grids 
group_var = ['Year', 'Month', 'Day']
df_day_JJASO_temp = df.groupby(group_var)[model_var].mean()
df_day_JJASO_temp = df_day_JJASO_temp.reset_index()
```

```python
## prepare legend handels, labels for each group and add legend to the plot
import matplotlib.patches as mpatches

var_list = ['APHRO','ERA-I', 'CCLM (EAS)', 'HIRHAM (EAS)', 'HadRM3P (EAS)', 'RegCM4-1 (WAS)', 'RegCM4-4 (WAS)', 'HadRM3P (WAS)',  'RCA4 (WAS)' ]
colors_model =  {'APHRO':'black', 
                 'ERA-I':'grey', 
                 'HIRHAM (EAS)': 'red', 
                 'HadRM3P (EAS)': 'blue', 
                 'RegCM4-1 (WAS)': 'lawngreen', 
                 'RegCM4-4 (WAS)': 'sienna', 
                 'HadRM3P (WAS)': 'orange', 
                 'RCA4 (WAS)': 'cyan',
                 'CCLM (EAS)': 'blueviolet'} #magenta  "blueviolet", 



legend_handles = [
    mpatches.Patch(color=colors_model[var_list[0]], label=var_list[0]),
    mpatches.Patch(color=colors_model[var_list[1]], label=var_list[1]), # add as many as needed
    mpatches.Patch(color=colors_model[var_list[2]], label=var_list[2]),
    mpatches.Patch(color=colors_model[var_list[3]], label=var_list[3]),
    mpatches.Patch(color=colors_model[var_list[4]], label=var_list[4]),
    mpatches.Patch(color=colors_model[var_list[5]], label=var_list[5]),
    mpatches.Patch(color=colors_model[var_list[6]], label=var_list[6]),
    mpatches.Patch(color=colors_model[var_list[7]], label=var_list[7]),
    mpatches.Patch(color=colors_model[var_list[8]], label=var_list[8]),
]
```

```python
####################################################################
### Fig 2A. scatter plot for precipitation
####################################################################
import matplotlib.pyplot as plt

# set df list
Var = [df_day_cord_preci, df_mon_cord_preci, df_yr_cord_preci,
       df_day_cord_JJASO_preci, df_mon_cord_JJASO_preci, df_yr_cord_JJASO_preci]
labels =['a) Day, All-S', 'b) Mon, All-S', 'c) Yr, All-S',
          'd) Day, JJASO', 'e) Mon, JJASO', 'f) Yr, JJASO']

# set plot parameters
size = 1.5
alpha = 0.4
marker = 'o'
width = 18
height = 27
xlim = [400, 40, 20, 400, 40, 20]  # xlim for each subplot

# set subplot parameters
figure, axis = plt.subplots(nrows=9, ncols=6, figsize=(width, height))#, sharex=True, sharey=True)
figure.tight_layout() # 

# plot subplots
for k in range(9): 
    for i in range(6):
        
        
        axis[k, i].scatter(Var[i][var_list[0]], Var[i][var_list[k]], 
                            marker=marker, s=size, alpha = alpha, c=colors_model[var_list[k]])
        
        if k==0:
            axis[k, i].set_title(labels[i], fontsize=22)
        axis[k, i].set_xlim(0, xlim[i])
        axis[k, i].set_ylim(0, xlim[i])
        axis[k, i].tick_params(axis='x', labelsize=16)
        axis[k, i].tick_params(axis='y', labelsize=16)
        axis[k, i].axline((0, 0), slope=1, linestyle='-', lw=0.2, color='grey')
        axis[k, i].set_aspect('equal', adjustable='box')
        axis[k, i].grid(True)

# For Sine Function

figure.text(0.5, -0.01, 'APHRO Precipitation (mm/day)', ha='center', fontsize=24)
figure.text(-0.03, 0.5, 'APHRO, ERA-I or RCM Precipitation (mm/day)', va='center', rotation='vertical', fontsize=24)
figure.legend(
    
    handles=legend_handles,          
    bbox_to_anchor=(1.004, -0.01), #bbox_to_anchor=(1, 0.97)
    bbox_transform=plt.gcf().transFigure,
    handlelength=1, handletextpad=0.5,
    title_fontsize=24,
    fontsize=18,
    ncol=7, 
    columnspacing=1.65)

plt.savefig(
    "../Charts/Fig2A_preci_scatterplot_3time-space.tiff", 
    bbox_inches='tight',
    dpi=300, 
    format="tiff", 
    pil_kwargs={"compression": "tiff_lzw"}
)
plt.show()

```

```python
###########################################################################################
### Fig 2B. scatter plot for temperature
###########################################################################################
import matplotlib.pyplot as plt

# set df list
Var = [df_day_cord_temp, df_mon_cord_temp, df_yr_cord_temp,
       df_day_cord_JJASO_temp, df_mon_cord_JJASO_temp, df_yr_cord_JJASO_temp]
labels = ['a) Day, All-S', 'b) Mon, All-S', 'c) Yr, All-S',
          'd) Day, JJASO', 'e) Mon, JJASO', 'f) Yr, JJASO']

# set plot parameters
size = 1.5
alpha = 0.4
marker = 'o'
width = 18
height = 24
xlim1 = [-30, -20, -20, -30, -20, -20]  # xlim for each subplot
xlim2 = [40, 40, 40, 40, 40, 40]  # xlim for each subplot

# set subplot parameters
figure, axis = plt.subplots(nrows=8, ncols=6, figsize=(width, height))#, sharex=True, sharey=True)
figure.tight_layout() # 

# plot subplots
for k in range(8):
    for i in range(6):
        if k < 6:
        
            axis[k, i].scatter(Var[i][var_list[0]], Var[i][var_list[k]], 
                                marker=marker, s=size, alpha = alpha, c=colors_model[var_list[k]])
        else:
            axis[k, i].scatter(Var[i][var_list[0]], Var[i][var_list[k+1]], 
                                marker=marker, s=size, alpha = alpha, c=colors_model[var_list[k+1]])
        axis[k, i].set_xlim(xlim1[i], xlim2[i])
        axis[k, i].set_ylim(xlim1[i], xlim2[i])
        axis[k, i].tick_params(axis='x', labelsize=16)
        axis[k, i].tick_params(axis='y', labelsize=16)
        axis[k, i].axline((0, 0), slope=1, linestyle='-', lw=0.2, color='grey')
        axis[k, i].set_aspect('equal', adjustable='box')
        axis[k, i].grid(True)
            
        if k==0:
            axis[k, i].set_title(labels[i], fontsize=22)
        


figure.text(0.5, -0.01, 'APHRO Temperature (°C)', ha='center', fontsize=24)
figure.text(-0.03, 0.5, 'APHRO, ERA-I or RCM Temperature (°C)', va='center', rotation='vertical', fontsize=24)
figure.legend(
    
    handles=legend_handles,          
    bbox_to_anchor=(1.004, -0.01), #bbox_to_anchor=(1, 0.97)
    bbox_transform=plt.gcf().transFigure,
    handlelength=1, handletextpad=0.5,
    title_fontsize=24,
    fontsize=18,
    ncol=7, #mode="expand", 
    columnspacing=1.65)

plt.savefig(
    "../Charts/Fig2B_temp_scatterplot_3time-space.tiff", 
    bbox_inches='tight',
    dpi=300, 
    format="tiff", 
    pil_kwargs={"compression": "tiff_lzw"}
)
plt.show()

```
